```python
import pandas as pd
```

```python
import random

random.seed(42)

# Sample data for generating realistic applicant records
names = [
    "Ahmed Mohamed", "Sara Ali", "Omar Hassan", "Mariam Adel",
    "Youssef Ibrahim", "Nour Khaled", "Salma Mahmoud",
    "Mostafa Samir", "Farah Tarek", "Amr Ashraf",
    "Hana Mostafa", "Karim Wael", "Aya Hossam", "Ali Gamal",
    "Malak Ahmed", "Ziad Sherif", "Nada Emad", "Khaled Fathy",
    "Jana Sameh", "Mahmoud Reda"
]

universities = [
    "Cairo University", "Ain Shams University","Helwan University", "Alexandria University",
    "Mansoura University", "BUE", "AUC",
    "Zagazig University", None, "Helwan Univ."
]

domains = [
    "Web Development", "web dev", "WEB DEV",
    "Data Science", "data science", "DATA SCIENCE",
    "AI", "Artificial Intelligence", "artificial intelligence",
    "Cyber Security", "cyber security", "Cybersecurity",
    "UI/UX", "ui/ux", "UI UX"
]

statuses = [
    "Under Review", "under review", "UNDER REVIEW",
    "Selected", "selected", "SELECTED",
    "Rejected", "rejected", "REJECTED", None
]

def random_date():
    year = random.choice([2025, 2026])
    month = random.randint(1, 12)
    day = random.randint(1, 28)

    date = pd.Timestamp(year=year, month=month, day=day)

    # Deliberately use different date formats
    if random.choice([True, False]):
         return date.strftime("%Y-%m-%d")
    return date.strftime("%d/%m/%Y")


records = []

for i in range(80):
    name = random.choice(names)
    email_name = name.lower().replace(" ", ".")
    email = f"{email_name}{i}@example.com"

    # Introduce missing values
    if random.random() < 0.07:
        name = None

    if random.random() < 0.10:
        email = None

    phone = f"01{random.randint(0, 2)}-{random.randint(10000000, 99999999)}"

    if random.random() < 0.15:
        phone = phone.replace("-", " ")

    if random.random() < 0.08:
        phone = None

    domain = random.choice(domains) 
    university = random.choice(universities)
    application_date = random_date()
    status = random.choice(statuses)

    # Introduce extra whitespace in some text values
    if name is not None and random.random() < 0.15:
        name = f" {name} "

    records.append({
        "Applicant Name": name,
        "Email": email,
        "Phone": phone,
        "Domain Applied": domain,
        "University": university,
        "Application Date": application_date,
        "Status": status
    })

df = pd.DataFrame(records)

# Ensure known duplicate examples exist
df.loc[8, "Applicant Name"] = "Ahmed Mohamed"
df.loc[8, "Email"] = "ahmed.duplicate@example.com"
df.loc[8, "Phone"] = "010-12345678"
df.loc[8, "Domain Applied"] = "Web Development"
df.loc[8, "University"] = "Cairo University"
df.loc[8, "Application Date"] = "2026-05-12"
df.loc[8, "Status"] = "Selected"

# Exact duplicate of row 9
df.iloc[9] = df.iloc[8].copy()

# Another application using the same email, with different details
df.loc[19, "Applicant Name"] = "Sara Ali"
df.loc[19, "Email"] = "sara.duplicate@example.com"
df.loc[19, "Phone"] = "010-87654321"
df.loc[19, "Domain Applied"] = "Data Science"
df.loc[19, "University"] = "Helwan University"
df.loc[19, "Application Date"] = "12/05/2026"
df.loc[19, "Status"] = "Under Review"

df.loc[20, "Applicant Name"] = " Sara Ali "
df.loc[20, "Email"] = "sara.duplicate@example.com"
df.loc[20, "Phone"] = "010 87654321"
df.loc[20, "Domain Applied"] = "data science"
df.loc[20, "University"] = "Helwan Univ."
df.loc[20, "Application Date"] = "2026-05-15"
df.loc[20, "Status"] = "under review"

# Save the raw dataset without the DataFrame index
df.to_csv("applicants.csv", index=False)

print("Dataset created successfully!")
print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))
print("Columns:", df.columns.tolist())
```

```python
# Count missing applicant names before handling them
print("Missing names before:", df["Applicant Name"].isnull().sum())

# Remove rows with missing applicant names
df = df.dropna(subset=["Applicant Name"])

print("Missing names after:", df["Applicant Name"].isnull().sum())
print("Dataset shape after dropping rows:", df.shape)
```

```python
# Count missing emails before handling them
print("Missing emails before:", df["Email"].isnull().sum())

# Fill missing email addresses with a placeholder
df["Email"] = df["Email"].fillna("not_provided@example.com")

print("Missing emails after:", df["Email"].isnull().sum())
```

```python
print("Missing universities before:", df["University"].isnull().sum())
df["University"] = df["University"].fillna("Not Specified")
print("Missing universities after:", df["University"].isnull().sum())
```

```python

print("Missing statuses before:", df["Status"].isnull().sum())
df["Status"] = df["Status"].fillna("Under Review")
print("Missing statuses after:", df["Status"].isnull().sum())
```

```python

print("Dataset shape:", df.shape)
print("\nMissing values after Step 4:")
print(df.isnull().sum())
print("\nTotal missing values:")
print(df.isnull().sum().sum())
```

```python

# Remove exact duplicate rows
df = df.drop_duplicates()

# Verify the result
print("Dataset shape after removing exact duplicates:", df.shape)
print("Remaining exact duplicates:", df.duplicated().sum())
```

```python

# Checking Duplicate Applications by Email
# I checked for duplicate applications using Email as an applicant
# identifier because the same applicant may submit multiple applications with slightly different details.
# I excluded the placeholder email "not_provided@example.com"
# because it does not identify a specific applicant.

# Exclude placeholder emails
valid_email = df["Email"] != "not_provided@example.com"

# Find all records with repeated valid email addresses
duplicate_emails = df.loc[
    valid_email & df["Email"].duplicated(keep=False)
].sort_values("Email")

print("Records with duplicate valid emails:", len(duplicate_emails))
print("Additional duplicate applications:",
      df.loc[valid_email, "Email"].duplicated().sum())

duplicate_emails
```

```python
# Remove duplicate applications based on valid emails
duplicate_mask = (
    df["Email"].ne("not_provided@example.com")
    & df["Email"].duplicated()
)

df = df.loc[~duplicate_mask].copy()

# Verify the results
print("Dataset shape:", df.shape)
print("Remaining exact duplicates:", df.duplicated().sum())
```

```python
df["Domain Applied"].unique()
```

```python

# Remove extra whitespace and standardize casing
df["Domain Applied"] = df["Domain Applied"].str.strip().str.lower()
```

```python
# Standardize domain names
domain_mapping = {
    "web dev": "Web Development", "web development": "Web Development",
    "data science": "Data Science",
    "ai": "Artificial Intelligence", "artificial intelligence": "Artificial Intelligence",
    "cyber security": "Cyber Security", "cybersecurity": "Cyber Security",
    "ui ux": "UI/UX", "ui/ux": "UI/UX"
}

df["Domain Applied"] = df["Domain Applied"].map(domain_mapping)

df["Domain Applied"].unique()
```

```python
df["Status"] = df["Status"].str.strip().str.title()
df["Status"].unique()
```

```python
# Remove extra whitespace from all text columns
for col in df.select_dtypes(include=["object", "string"]).columns:
    df[col] = df[col].str.strip()
```

```python

print("Domains:", df["Domain Applied"].nunique())
print("Statuses:", df["Status"].unique())
print("Missing values:\n", df.isnull().sum())
```

```python

df["Application Date"] = pd.to_datetime(
    df["Application Date"], errors="coerce", dayfirst=True
)

df["Application Date"].isnull().sum()
```

```python

# Restore original dates for the remaining rows
raw_dates = pd.read_csv("applicants.csv")["Application Date"]
df["Application Date"] = raw_dates.loc[df.index]

# Parse mixed date formats
df["Application Date"] = pd.to_datetime(
    df["Application Date"], format="mixed",
    dayfirst=True, errors="coerce"
)

print("Missing dates:", df["Application Date"].isnull().sum())
print("Date type:", df["Application Date"].dtype)
```

```python
df["Phone"] = df["Phone"].str.replace(r"[-\s]", "", regex=True)
df["Phone"].head()
```
